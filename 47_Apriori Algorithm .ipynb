# Machine Learning 47: Apriori Algorithm

## What is the Apriori Algorithm?

The **Apriori Algorithm** is a **classic algorithm in data mining** used for **association rule learning** — that is, finding relationships or patterns among items in large datasets (especially transactional data).

It’s mainly used in **Market Basket Analysis** — for example:

> "If a customer buys bread and butter, they are likely to buy milk."


## Purpose

The goal of the Apriori algorithm is to find **frequent itemsets** (items that appear together often) and then **derive association rules** from those itemsets.


## How It Works (Step-by-Step)

### 1️. Set Minimum Thresholds

You define two thresholds:
- **Minimum Support:** how often an itemset appears in transactions.
- **Minimum Confidence:** how often the rule is found to be true.

### 2️. Find Frequent Itemsets

The algorithm goes level by level:
- Find all 1-itemsets (individual items) that meet the **minimum support**.
- Use them to generate 2-itemsets, then 3-itemsets, and so on.
- Continue until no more frequent itemsets are found.

### 3️. Generate Association Rules

From the frequent itemsets, the algorithm generates rules that meet **minimum confidence**.  
For example, if `{milk, bread}` → `{butter}` has high confidence, it means:

> People who buy milk and bread are likely to buy butter too.


##  Key Terms

| Term | Meaning |
|------|----------|
| **Support(X)** | Probability that transaction contains itemset X |
| **Confidence(X→Y)** | Probability that Y is purchased given X is purchased |
| **Lift(X→Y)** | How much more likely Y is bought when X is bought compared to chance |


##  Mathematical Formulas

$$
Support(X \rightarrow Y) = \frac{\text{Transactions containing both } X \text{ and } Y}{\text{Total number of transactions}}
$$

$$
Confidence(X \rightarrow Y) = \frac{Support(X \cup Y)}{Support(X)}
$$

$$
Lift(X \rightarrow Y) = \frac{Confidence(X \rightarrow Y)}{Support(Y)}
$$

## Applications

- Market basket analysis  
- Recommender systems (e.g., Amazon: “Customers also bought…”)  
- Web usage mining

```python
import pandas as pd
import numpy as np
```

```python
# Load the dataset
dataset = pd.read_csv("groceries - groceries.csv")
```

```python
dataset.head()
```

```python
# Check the shape of the dataset
dataset.shape
```

```python
# Create the market basket list 
market = []
for i in range(0, dataset.shape[0]):
    cus = []
    for j in dataset.columns:
        if type(dataset[j][i]) == str:
            cus.append(dataset[j][i])
    market.append(cus)
```

```python
# Flatten the market list
l = []
for i in market:
    for j in i:
        l.append(j)
```

```python
l
```

```python
import collections
```

```python
# Count the frequency of each item in the list
p = collections.Counter(l) 
```

```python
d = {"Item Name": list(p.keys()), "Frequency": list(p.values())}
```

```python
# Create a dataframe to display the frequency of each item
pd.DataFrame(d).sort_values(by = "Frequency", ascending = False)
```

```python
df =pd.DataFrame(tr.transform(market),columns=tr.columns_)
```
